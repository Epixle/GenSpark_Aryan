```python
# Create 5 default books as dictionaries

book1 = {
    "title": "Catch-22",
    "author": "Joseph Heller",
    "year": 1961,
    "genre": "Comedy"
}

book2 = {
    "title": "To Kill A Mockingbird",
    "author": "Harper Lee",
    "year": 1960,
    "genre": "Southern Gothic"
}

book3 = {
    "title": "Don Quixote",
    "author": "Miguel de Cervantes",
    "year": 1605,
    "genre": "Comedy"
}

book4 = {
    "title": "1984",
    "author": "George Orwell",
    "year": 1949,
    "genre": "Dystopian"
}

book5 = {
    "title": "Animal Farm",
    "author": "George Orwell",
    "year": 1945,
    "genre": "Dystopian"
}
```

```python
# Add all the books to a list called library
library = [book1, book2, book3, book4, book5]

print(library)
```

```python
# Create a set to store book title and author. Initialize with one book to turn to set
bookInfo = {("Catch-22", "Joseph Heller")}

# For each book, turn the title and author name into a tuple and put into set
for book in library:
    bookInfo.add((book.get("title"), book.get("author")))

print(bookInfo)
```

```python
# Create a set to store unique book genres. Initialize with one genre to turn to set
bookGenres = {"Comedy"}

# For each book, add the genre into set
for book in library:
    bookGenres.add(book.get("genre"))

print(bookGenres)
```

```python
# Validates user input
def validate(str, num = False):
    # Keep asking for an input until it is valid
    while True:
        # Get user input
        answer = input(str)
        
        # If the input is blank, make user input again
        if not answer:
            print("Input is not valid. Please try again")
            continue
        
        # If the input is supposed to be a number, check to see if it is
        if num:
            if answer.isdigit():
                answer = int(answer)
            else:
                # If not a number, ask user for input
                print("Input is not valid. Please try again.")
                continue

        return answer
```

```python
# Add a book to the list
def addBook():
    title = validate("What is the title of the book?: ").title()
    author = validate("What is the name of the author?: ").title()

    # Cross-references bookInfo list to see if book already exists in library
    if (title, author) in bookInfo:
        print("This book is already in the library.")
        return
    
    year = validate("What year was the book published?: ", True)
    genre = validate("What genre is the book?: ").title()

    # Add book to library, title and author tuple to bookInfo, and genre to bookGenre
    library.append({"title": title, "author": author, "year": year, "genre": genre})
    bookInfo.add((title, author))
    bookGenres.add(genre)

    print(f"Added {title} ({year}) by {author} to the library.")
```

```python
# List every book in library
def viewBooks():
    print("The books in your library are:")
    
    # Get every title and author tuple in bookInfo set and print it
    for book in bookInfo:
        print(book[0], "by", book[1])
```

```python
# Search for a specific book according to its title
def search():
    title = validate("Which book are you looking for?: ").title()

    # Iterate through every book in the library
    for book in library:
        # If search matches any book title, list the book details
        if book.get("title") == title:
            print(f'{book.get("title")} ({book.get("year")}) by {book.get("author")}')
            print("Genre:", book.get("genre"))
            break
```

```python
# List every book according to genre user is looking for
def filterGenre():
    genre = validate("What genre are you looking for?: ").title()

    # Iterate through every book in the library
    for book in library:
        # If book genre matches what user is looking for, print details and keep looking
        if book.get("genre") == genre:
            print(f'{book.get("title")} ({book.get("year")}) by {book.get("author")}')
```

```python
# List the unique genres in the library
def uniqueGenres():
    print("The different genres available in the library are:")

    # Iterate through bookGenres set and print out every genre
    for genre in bookGenres:
        print(genre)
```

```python
decision = 0

# Ask user for their decision until they exit
while decision != 6:
    decision = input('''What would you like to do?:
    1. Add a book
    2. View all books
    3. Search by title
    4. Filter by genre
    5. Show unique genres
    6. Exit
''')

    # Call function depending on input
    match decision:
        case "1":
            addBook()
        case "2":
            viewBooks()
        case "3":
            search()
        case "4":
            filterGenre()
        case "5":
            uniqueGenres()
        case "6":
            break
        case _:
            print("Input is not valid. Please try again")
    
    #buffer
    print()
```
